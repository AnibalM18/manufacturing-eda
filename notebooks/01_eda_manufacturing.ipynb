# Exploratory Data Analysis: Machine Failures in Manufacturing

**Dataset:** AI4I 2020 Predictive Maintenance (UCI Machine Learning Repository).
Note: this is a *synthetic* dataset designed to reflect real predictive
maintenance data, not measurements from a physical plant.

**Goal:** identify which operating conditions are associated with machine failures.

**Tools:** pandas, matplotlib, seaborn, scikit-learn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Consistent visual style for all charts and show all columns when displaying DataFrames
sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
```

```python
# Load the raw data 
df = pd.read_csv("../data/raw/ai4i2020.csv")
df.shape
```

```python
# Basic information about data
print(df.head())
print(df.info())
print(df.describe())
print(df.isna().sum())
print(df.duplicated().sum())
```

```python
# Class balance of the target variable. Counts and proportions
print(df["Machine failure"].value_counts())
print(df["Machine failure"].value_counts(normalize=True))
```

**Observation:** only 3.39% of products experienced a failure, so the dataset is highly imbalanced. Accuracy alone would be a misleading metric for any predictive model.
